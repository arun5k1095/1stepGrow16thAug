```python
# DEFINING FUNCTIONS: Function with no inputs
# A simple function that returns a greeting message
def greet():
    return "Hello, World!"

# CALLING THE FUNCTION: Execute the function multiple times
print("Calling greet() function:")
print(greet())
print(greet())
print("Notice: Same output every time since no inputs change the behavior")
```

```python
# DEFINING FUNCTIONS: Function with single input
# This function takes one parameter and returns its square
def square(number):
    return number * number

# CALLING THE FUNCTION: Pass different values as input
print("Calculating squares of different numbers:")
print(f"Square of 5 is: {square(5)}")
print(f"Square of 12 is: {square(12)}")
print(f"Square of 3.5 is: {square(3.5)}")
print("Notice: Different inputs produce different outputs")
```

```python
# DEFINING FUNCTIONS: Function with multiple inputs
# This function calculates the area of a rectangle using length and width
def calculate_rectangle_area(length, width):
    area = length * width
    return area

# CALLING THE FUNCTION: Pass different length and width values
print("Calculating areas of different rectangles:")
print(f"Rectangle (10 x 5): {calculate_rectangle_area(10, 5)} square units")
print(f"Rectangle (7.5 x 4.2): {calculate_rectangle_area(7.5, 4.2)} square units")
print(f"Rectangle (15 x 8): {calculate_rectangle_area(15, 8)} square units")
print("Notice: Each call uses different input values")
```

```python
# DEFINING FUNCTIONS: Function that processes strings
# This function creates an email address from first name, last name, and domain
def create_email(first_name, last_name, domain):
    email = first_name.lower() + "." + last_name.lower() + "@" + domain
    return email

# CALLING THE FUNCTION: Create different email addresses
print("Creating email addresses:")
print(f"Email 1: {create_email('John', 'Doe', 'company.com')}")
print(f"Email 2: {create_email('Alice', 'Smith', 'university.edu')}")
print(f"Email 3: {create_email('Bob', 'Johnson', 'startup.io')}")
print("Notice: String manipulation inside the function")
```

```python
# DEFINING FUNCTIONS: Function with mathematical operations
# This function converts temperature from Fahrenheit to Celsius
def convert_temperature(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius

# CALLING THE FUNCTION: Convert different temperatures
print("Converting temperatures from Fahrenheit to Celsius:")
print("32°F =", convert_temperature(32), "°C")
print("100°F =", convert_temperature(100), "°C")
print("68°F =", convert_temperature(68), "°C")
print("Notice: Mathematical calculations inside the function")
```

```python
# DEFINING FUNCTIONS: Function that works with lists
# This function finds the maximum value in a list of numbers
def find_maximum(numbers):
    max_value = numbers[0]
    for num in numbers:
        if num > max_value:
            max_value = num
    return max_value

# CALLING THE FUNCTION: Find maximum in different lists
print("Finding maximum values in lists:")
print("Max of [3, 7, 2, 9, 1]:", find_maximum([3, 7, 2, 9, 1]))
print("Max of [15, 8, 23, 4]:", find_maximum([15, 8, 23, 4]))
print("Max of [100, 50, 75]:", find_maximum([100, 50, 75]))
print("Notice: Function processes a list and returns single value")
```

```python
# DEFINING FUNCTIONS: Function with NO return statement
# This function prints a personalized birthday message
def print_birthday_message(name, age):
    print("Happy Birthday", name + "!")
    print("You are now", age, "years old!")
    print("Have a wonderful day!")

# CALLING THE FUNCTION: Print messages for different people
print("Calling print_birthday_message() function:")
print_birthday_message("Alice", 25)
print()  # Empty line for spacing
print_birthday_message("Bob", 30)
print("Notice: This function prints directly, no return value")
```

```python
# DEFINING FUNCTIONS: Function with complex calculation
# This function calculates compound interest
def calculate_compound_interest(principal, rate, time, compounds_per_year):
    amount = principal * (1 + rate / compounds_per_year) ** (compounds_per_year * time)
    interest = amount - principal
    return interest

# CALLING THE FUNCTION: Calculate interest for different scenarios
print("Calculating compound interest:")
print("$1000 at 5% for 2 years compounded quarterly:", calculate_compound_interest(1000, 0.05, 2, 4))
print("$5000 at 7% for 5 years compounded monthly:", calculate_compound_interest(5000, 0.07, 5, 12))
print("$2000 at 3% for 10 years compounded annually:", calculate_compound_interest(2000, 0.03, 10, 1))
print("Notice: Multiple parameters working together in calculation")
```

```python
# DEFINING FUNCTIONS: Function that counts characters
# This function counts vowels in a string
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count = count + 1
    return count

# CALLING THE FUNCTION: Count vowels in different words
print("Counting vowels in words:")
print("Word 'hello' has", count_vowels("hello"), "vowels")
print("Word 'programming' has", count_vowels("programming"), "vowels")
print("Word 'education' has", count_vowels("education"), "vowels")
print("Notice: Function processes each character in the string")
```

```python
# DEFINING FUNCTIONS: Function that checks conditions
# This function determines if a number is even or odd
def check_even_odd(number):
    if number % 2 == 0:
        return "even"
    else:
        return "odd"

# CALLING THE FUNCTION: Check different numbers
print("Checking if numbers are even or odd:")
print("Number 10 is", check_even_odd(10))
print("Number 7 is", check_even_odd(7))
print("Number 22 is", check_even_odd(22))
print("Notice: Function uses conditional logic to return different values")
```

```python
# DEFINING FUNCTIONS: Function that calculates age
# This function calculates age given birth year
def calculate_age(birth_year, current_year):
    age = current_year - birth_year
    return age

# CALLING THE FUNCTION: Calculate ages for different birth years
print("Calculating ages:")
print("Born in 1990, current year 2024:", calculate_age(1990, 2024), "years old")
print("Born in 2000, current year 2024:", calculate_age(2000, 2024), "years old")
print("Born in 1985, current year 2024:", calculate_age(1985, 2024), "years old")
print("Notice: Simple subtraction calculation")
```

```python
# DEFINING FUNCTIONS: Function that prints pattern (NO RETURN)
# This function prints a simple star pattern
def print_star_pattern(rows):
    for i in range(1, rows + 1):
        print("*" * i)

# CALLING THE FUNCTION: Print different sized patterns
print("Printing star patterns:")
print("Pattern with 3 rows:")
print_star_pattern(3)
print("\nPattern with 5 rows:")
print_star_pattern(5)
print("Notice: Function only prints, doesn't return anything")
```

```python
# DEFINING FUNCTIONS: Function that calculates sum of list
# This function calculates the sum of all numbers in a list
def calculate_sum(numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total

# CALLING THE FUNCTION: Calculate sum of different lists
print("Calculating sum of numbers in lists:")
print("Sum of [1, 2, 3, 4, 5]:", calculate_sum([1, 2, 3, 4, 5]))
print("Sum of [10, 20, 30]:", calculate_sum([10, 20, 30]))
print("Sum of [7, 14, 21, 28]:", calculate_sum([7, 14, 21, 28]))
print("Notice: Function iterates through list and accumulates values")
```

```python
# DEFINING FUNCTIONS: Function that converts text case
# This function converts text to title case
def make_title_case(text):
    words = text.split()
    title_words = []
    for word in words:
        title_word = word[0].upper() + word[1:].lower()
        title_words.append(title_word)
    result = " ".join(title_words)
    return result

# CALLING THE FUNCTION: Convert different text strings
print("Converting text to title case:")
print("hello world becomes:", make_title_case("hello world"))
print("PYTHON PROGRAMMING becomes:", make_title_case("PYTHON PROGRAMMING"))
print("mIxEd CaSe becomes:", make_title_case("mIxEd CaSe"))
print("Notice: Function processes each word in the string")
```
