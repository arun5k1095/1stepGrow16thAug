Collections in Python<br> <ul>
<li><strong><a href="python_lists.asp">List</a></strong> is a collection which is ordered and changeable. Allows duplicate members.</li>
<li><strong>Tuple</strong> is a collection which is ordered and unchangeable. Allows duplicate members.</li>
<li><strong><a href="python_sets.asp">Set</a></strong> is a collection which is unordered, 
unchangeable*, and unindexed. No duplicate members.</li>
<li><strong><a href="python_dictionaries.asp">Dictionary</a></strong> is a collection which is ordered** 
and changeable. No duplicate members.</li>
</ul><b> LIST </b>


```python
list1 = ["abc", 34, True, 40, "male"]
print(type(list1))

thislist = list(("apple", "banana", "cherry")) # note the double round-brackets
print(thislist)

print(thislist[1])
print(thislist[-1])

thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:5])

# check item in list
thislist = ["apple", "banana", "cherry"]
if "apple" in thislist:
  print("Yes, 'apple' is in the fruits list")

#insert in a list
thislist.insert(2, "watermelon")
print(thislist)

#Add item in list
thislist.append("orange")
print(thislist)


#AddRange join 2 list
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
thislist.extend(tropical)
print(thislist)

# Note : extend() method does not have to append lists, you can add any iterable object (tuples, sets, dictionaries
thislist = ["apple", "banana", "cherry"]
thistuple = ("kiwi", "orange")
thislist.extend(thistuple)
print(thislist)

thislist = ["apple", "banana", "cherry"]
thislist.remove("banana")  #the remove() method removes the first occurrence
print(thislist)


## pop
thislist = ["apple", "banana", "cherry"]
thislist.pop(1)
print(thislist)

##pop() will remove the last item

thislist = ["apple", "banana", "cherry"]
thislist.pop()
print(thislist)

# clear the list

thislist = ["apple", "banana", "cherry"]
thislist.clear()
print(thislist)
```

<h2>Tuple</h2>
<p>Tuples are used to store multiple items in a single variable.</p>

```python
tuple1 = ("abc", 34, True, 40, "male")
print(tuple1)
print(len(tuple1))
print(type(tuple1))

#Trick question - One Item tuple
thistuple = ("apple",)
print(len(thistuple))
print(type(tuple1))


#ACCESS TUPLE
print(tuple1[1])

# tuples are immutable
#   thistuple[0] = "orange"      ## TypeError: 'tuple' object does not support item assignment

#workaround - create a new tuple and assign
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y
print(thistuple)
```
